Si vous modifiez `solution_results.py`, redémarrez le kernel avant de relancer les tests.

```python
import sys
from pathlib import Path

import pytest
from returns.result import Failure, Success

HERE = Path.cwd()  # le notebook est dans le dossier results/
if str(HERE) not in sys.path:
    sys.path.insert(0, str(HERE))

from solution_results import parse_user_input

print(sys.executable)
```

## 1. Cas valides

```python
for name, age, email in [
    ("Alice", 30, "alice@example.com"),
    ("  Bob  ", 25, "bob@example.com"),
]:
    result = parse_user_input(name=name, age=age, email=email)
    print(repr(name), "->", result)
```

## 2. Cas invalides

```python
cases = [
    ("", 30, "alice@example.com"),
    ("x" * 101, 30, "x@example.com"),
    ("Alice", -1, "alice@example.com"),
    ("Alice", 151, "alice@example.com"),
    ("Alice", 30, "notanemail"),
]
for name, age, email in cases:
    result = parse_user_input(name=name, age=age, email=email)
    print(f"{name[:12]!r:16} age={age:<4} email={email!r:22} -> {result}")
```
